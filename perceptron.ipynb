```python
import numpy as np
```

```python
X=np.array([[1,0],[0,1],[1,1],[0,0]])
```

```python
y_or=np.array([1,1,1,0])
```

```python
def step(z):
  return (z>0).astype(int)
def Forward_pass(X,w,b):
 z=X@w+b
 return step(z)
```

```python
w=np.array([0,0])
b=0
```

```python
y_pred=Forward_pass(X,w,b)
```

```python
print("y_pred",y_pred)
print("y",y_or)
```

```python
def compute_loss(y,y_pred):
  return y-y_pred
loss=compute_loss(y_or,y_pred)
print("A\tB\tActual\tPredicted\tLoss")

for i in range(len(X)):
    print(
        X[i][0], "\t",
        X[i][1], "\t",
        y_or[i], "\t",
        y_pred[i], "\t\t",
        loss[i]
    )
```

```python
def perceptron(w,b,loss,X):
  total_loss=0
  for i in range(len(X)):
    total_loss+=abs(loss)
    w=w+loss[i]*X[i]
    b=b+loss[i]
  return w,b
```

```python
perceptron(w,b,loss,X)
```

```python
def train_perceptron(X, y, w, b, epochs=20):

    for epoch in range(epochs):

        y_pred = []

        for i in range(len(X)):
            z = np.dot(X[i], w) + b

            if z >= 0:
                y_pred.append(1)
            else:
                y_pred.append(0)

        y_pred = np.array(y_pred)

        loss = y - y_pred

        old_w = w.copy()
        old_b = b

        w, b = perceptron(w, b, loss, X)

        print("Epoch:", epoch + 1)
        print("Weight:", w)
        print("Bias:", b)
        print("Loss:", loss)
        print()

        if np.array_equal(old_w, w) and old_b == b:
            print("No update in weights and bias.")
            print("Training stopped.")
            break

    return w, b
```

```python
w, b = train_perceptron(X, y_or, w, b, 20)
```

```python
y_pred = Forward_pass(X, w, b)

print("Predicted y:", y_pred)
```

```python
accuracy = np.mean(y_pred == y_or)

print("Accuracy:", accuracy)
print("Accuracy %:", accuracy * 100)
```
